## II> Lọc Tuyến Tính
1. Lọc trung bình: Làm mờ ảnh bằng cách tính giá trị trung bình của các điểm ảnh trong một vùng lân cận.
2. Lọc Gaussian: Làm mờ ảnh một cách tự nhiên hơn bằng cách sử dụng kernel Gaussian.
3. Làm sắc nét: Tăng cường các cạnh trong ảnh bằng cách sử dụng các kernel thích hợp.


```python
from pathlib import Path
import cv2
import numpy as np
import matplotlib.pyplot as plt

BASE_DIR = Path(".").resolve()

IMAGE_PATH = BASE_DIR / "images" / "input.jpg"

OUTPUT_DIR = BASE_DIR / "output" / "bai2"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

img_original = cv2.imread(str(IMAGE_PATH))

if img_original is None:
    raise FileNotFoundError(f"Không tìm thấy ảnh: {IMAGE_PATH}")
```

# 1. Lọc trung bình: 

```python
average = cv2.blur(img_original, (5, 5))

cv2.imwrite(str(OUTPUT_DIR / "average.jpg"), average)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(average, cv2.COLOR_BGR2RGB))
plt.title("Average Filter")
plt.axis("off")

plt.show()
```

## 2. Lọc Gaussian

```python
gaussian = cv2.GaussianBlur(img_original, (5,5), 0)

cv2.imwrite(str(OUTPUT_DIR / "gaussian.jpg"), gaussian)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Filter")
plt.axis("off")

plt.show()
```

## 3. Làm sắc nét

```python
kernel = np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])

sharpen = cv2.filter2D(img_original, -1, kernel)

cv2.imwrite(str(OUTPUT_DIR / "sharpen.jpg"), sharpen)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(sharpen, cv2.COLOR_BGR2RGB))
plt.title("Sharpen")
plt.axis("off")

plt.show()
```
